## 简单函数

- `say_hi` 不需要参数，调用时直接打印问候语。
- `goat_pad` 接收一个字符串，并在它前后拼接 `goat` 后打印。
- `print_n` 使用循环重复打印字符串，每次打印占一行。

```python
def say_hi():
    print("Hello, world!")


def goat_pad(text):
    print(f"goat{text}goat")


def print_n(s, n):
    for _ in range(n):
        print(s)


say_hi()
goat_pad("hello")
print_n("Python", 3)
```

## 欧几里得算法

`gcd` 反复用较小数替换为两数相除的余数，直到余数为 0；此时另一个数就是最大公约数。函数先对输入取绝对值，因此一个或两个参数为负数时仍会返回非负的 GCD，这种行为有意义，负数与其绝对值有相同的约数，因此通常把最大公约数定义为非负数。按常见约定，`gcd(0, 0)` 返回 `0`。

```python
def gcd(m, n):
    """Return the non-negative greatest common divisor of two integers."""
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m


pairs = [(2018, 2019), (1200, 300), (5040, 60)]
for m, n in pairs:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")

print(f"gcd(-24, 18) = {gcd(-24, 18)}")
print(f"gcd(-24, -18) = {gcd(-24, -18)}")

assert [gcd(m, n) for m, n in pairs] == [1, 300, 60]
assert gcd(-24, 18) == 6
assert gcd(-24, -18) == 6
assert gcd(0, 0) == 0
```

## 近似欧拉数 e



```python
import math


def euler_limit(n):
    return (1 + 1 / n) ** n


def euler_infinite_sum(n):
    total = 0.0
    term = 1.0
    for k in range(n):
        total += term
        term /= k + 1
    return total


def euler_approx(epsilon):
    target = math.exp(1)
    total = 0.0
    term = 1.0
    k = 0
    while abs(target - total) >= epsilon:
        total += term
        k += 1
        term /= k
    return total


def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))


print("euler_limit(1000) =", euler_limit(1000))
print("euler_infinite_sum(4) =", euler_infinite_sum(4))
print("euler_approx(0.001) =", euler_approx(0.001))
print("euler sum table through 5:")
print_euler_sum_table(5)

assert euler_infinite_sum(0) == 0.0
assert euler_infinite_sum(1) == 1.0
assert math.isclose(euler_infinite_sum(4), 8 / 3)
assert abs(euler_approx(0.001) - math.exp(1)) < 0.001
```

## 测试整数属性

对于正整数，反复检查余数是否为 0；若能整除，就用整除结果继续检查。最后结果为 `1` 时，原数恰好是底数的整数次幂。`is_power_of_2` 只检查底数 `2`，不使用平方根；`is_power(b, n)` 检查一般正整数底数。按 $b^0=1$，任何正底数的 `1` 都是它的幂；底数为 `1` 时只有 `1` 是其幂。

```python
def is_power_of_2(n):
    while n % 2 == 0:
        n //= 2
    return n == 1


def is_power(b, n):
    if b == 1:
        return n == 1
    while n % b == 0:
        n //= b
    return n == 1


for value in [1, 2, 4, 8, 16]:
    print(f"is_power_of_2({value}) = {is_power_of_2(value)}")
print(f"is_power_of_2(12) = {is_power_of_2(12)}")

for base, value in [(2, 16), (3, 81), (1, 1)]:
    print(f"is_power({base}, {value}) = {is_power(base, value)}")
for base, value in [(2, 12), (3, 10), (1, 2)]:
    print(f"is_power({base}, {value}) = {is_power(base, value)}")

assert all(is_power_of_2(value) for value in [1, 2, 4, 8, 16])
assert not is_power_of_2(12)
assert is_power(2, 16)
assert is_power(3, 81)
assert is_power(5, 1)
assert is_power(1, 1)
assert not is_power(2, 12)
assert not is_power(3, 10)
assert not is_power(1, 2)
```
