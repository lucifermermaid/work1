# 一、字符串

 #### 1.回文是指一个词或短语，它从前往后读和从后往前读都一样

```python
def is_palindrome(str):
    normalized =str .replace(" ", "").lower()
    return normalized == normalized[::-1]


print(is_palindrome("pop"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("level"))
print(is_palindrome("hello"))
print(is_palindrome("kayak"))
print(is_palindrome("rats live on no evil star"))
print(is_palindrome("was it a car or a cat I saw"))
print(is_palindrome("hello"))
```

 #### 2.“字⺟顺序的”测试：即如果⼀个词的字⺟从左到右按字⺟顺序排列，就称它是“字⺟顺序的”；并且允许字⺟重复。

```python
def is_abecedarian(word):
    letters = word.replace(" ", "").lower()
    return all(
        letters[index] <= letters[index + 1]
        for index in range(len(letters) - 1)
    )


print(is_abecedarian("adder"))
print(is_abecedarian("beet"))
print(is_abecedarian("dog"))
print(is_abecedarian("cat"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("lucifer"))
```

#### 3.编写⼀个名为 remove_vowels 的函数，它接受⼀个字符串作为唯⼀参数，并返回去掉所有元⾳后的字符串。本题的元⾳是字⺟ a、e、i、o、u。


```python
def remove_vowels(text):
    vowels = "aeiou"
    return "".join(
        character for character in text
        if character.lower() not in vowels
    )


print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
print(remove_vowels("AEIOU"))
print(remove_vowels("mermaid"))
print(remove_vowels("xiaozhan"))
print(remove_vowels("yee"))
```

## 字符串练习结果

- `is_palindrome` 忽略空格和大小写；像 `T A C O cat` 这样的输入会判断为回文。
- `is_abecedarian` 忽略空格后检查字母是否按升序排列，重复字母允许出现。
- `remove_vowels` 会移除大小写形式的 `a`、`e`、`i`、`o`、`u`；全部是元音时返回空字符串。

# 二、列表

#### 1.编写⼀个名为 list_reverse 的函数，它接受⼀个列表作为参数，并返回反转后的列表。

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表")
    return lst[::-1]


value = [1, 2, 3] 
value2 = {'name':'lucifer'}

try:
    print(list_reverse(value))
except TypeError as error:
    print("类型错误：", error)
try:
    print(list_reverse(value2))
except TypeError as error:
    print("类型错误：", error)
```

#### 2.编写⼀个名为 is_sorted 的函数，它接受⼀个序列 seq 作为唯⼀参数。如果序列按升序排列，返回 True，否则返回 False。

```python
def is_sorted(seq):
    for index in range(1, len(seq)):
        if seq[index] < seq[index - 1]:
            return False
    return True


print(is_sorted([1, 2, 3]))
print(is_sorted([1, 1, 2]))
print(is_sorted([1, 3, 2]))
print(is_sorted([]))
print(is_sorted("abcd"))
print(is_sorted("abcdges"))
print(is_sorted("los"))
```

## 列表练习结果

- `list_reverse` 返回原列表元素顺序相反的新列表；非列表参数会引发 `TypeError`。
- `is_sorted` 逐对比较相邻元素，若没有出现前一个元素大于后一个元素的情况，就返回 `True`；空序列也视为升序。

# 三、作为向量的元组

在这个问题中，我们使用元组表示向量。这里用一个长度为 $d$ 的元组表示 $d$ 维向量。

#### 1. 实现名为 `vec_scalar_mult` 的函数

函数接收两个参数：由整数或浮点数组成的元组 `t`，以及整数或浮点数 `s`。返回与 `t` 长度相同的元组，其中每个元素等于 `t` 对应元素乘以 `s`。输入类型不符合要求时，引发带有适当错误信息的 `TypeError`。整数标量以及包含整数元素的元组都是有效输入。

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("t must be a tuple")
    if any(type(value) not in (int, float) for value in t):
        raise TypeError("all elements of t must be integers or floats")
    if type(s) not in (int, float):
        raise TypeError("s must be an integer or float")
    return tuple(value * s for value in t)


print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, 2, 3.5), 0.5))
print(vec_scalar_mult((4, 5), 3))

try:
    print(vec_scalar_mult([1, 2, 3], 2))
except TypeError as error:
    print("类型错误：", error)

try:
    print(vec_scalar_mult((1, "two", 3), 2))
except TypeError as error:
    print("类型错误：", error)

try:
    print(vec_scalar_mult((1, 2, 3), "2"))
except TypeError as error:
    print("类型错误：", error)
```

#### 2. 实现名为 `vec_inner_product` 的函数。函数接收两个由整数或浮点数组成的元组，返回这两个向量内积对应的浮点数。对于 $\mathbf{x}, \mathbf{y} \in \mathbb{R}^d$，内积定义为：$$\langle \mathbf{x}, \mathbf{y} \rangle = \sum_{j=1}^{d} x_j y_j$$若任一输入不是元组，应引发 `TypeError`；若两个元组长度不一致，应引发 `ValueError`。

```python
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("both inputs must be tuples")
    if any(type(value) not in (int, float) for value in t1):
        raise TypeError("all elements of t1 must be integers or floats")
    if any(type(value) not in (int, float) for value in t2):
        raise TypeError("all elements of t2 must be integers or floats")
    if len(t1) != len(t2):
        raise ValueError("both tuples must have the same length")
    return sum(float(a) * float(b) for a, b in zip(t1, t2))


print(vec_inner_product((1, 2, 3), (4, 5, 6)))
print(vec_inner_product((1.5, 2), (2, 0.5)))

try:
    print(vec_inner_product([1, 2], (3, 4)))
except TypeError as error:
    print("类型错误：", error)

try:
    print(vec_inner_product((1, 2), (3,)))
except ValueError as error:
    print("长度错误：", error)

try:
    print(vec_inner_product((1, "two"), (3, 4)))
except TypeError as error:
    print("类型错误：", error)
```

#### 3. 将向量表示方法扩展到矩阵

用一个元组的元组表示 $m \times n$ 矩阵 `my_mx`：外层元组长度为 $m$，每个内层元组长度为 $n$，表示矩阵的一行。编写名为 `check_valid_mx` 的函数，它接收一个参数；若参数是有效矩阵则返回 `True`，否则返回 `False`。有效矩阵需满足：

- 外层元组的每个元素本身都是元组；
- 所有内层元组长度相同；
- 每个内层元组中的元素都是整数或浮点数。

```python
def check_valid_mx(my_mx):
    if not isinstance(my_mx, tuple):
        return False
    if any(not isinstance(row, tuple) for row in my_mx):
        return False
    if my_mx and any(len(row) != len(my_mx[0]) for row in my_mx):
        return False
    if any(
        type(value) not in (int, float)
        for row in my_mx
        for value in row
    ):
        return False
    return True


print(check_valid_mx(((1, 2), (3.5, 4))))
print(check_valid_mx(((1, 2), (3,))))
print(check_valid_mx(((1, 2), [3, 4])))
print(check_valid_mx(((1, "x"), (3, 4))))
print(check_valid_mx(()))
```

#### 4. 编写名为 `mx_vec_mult` 的函数

函数接收一个矩阵（元组的元组）和一个向量（元组），返回矩阵与向量相乘得到的数值元组。矩阵应通过 `check_valid_mx` 检查；向量元素应为整数或浮点数，且向量长度必须等于矩阵的列数。输入类型不正确时引发 `TypeError`，维度不匹配时引发 `ValueError`。

## 矩阵检查结果

`check_valid_mx` 仅当外层和每一行都是元组、每行长度一致，且所有元素都是整数或浮点数时返回 `True`；其他输入返回 `False`。示例中的不规则行、列表行和字符串元素因此判为无效。

```python
def mx_vec_mult(my_mx, vector):
    if not check_valid_mx(my_mx):
        raise TypeError("my_mx must be a valid matrix of tuples")
    if not isinstance(vector, tuple):
        raise TypeError("vector must be a tuple")
    if any(type(value) not in (int, float) for value in vector):
        raise TypeError("all vector elements must be integers or floats")
    if not my_mx:
        return ()
    if len(vector) != len(my_mx[0]):
        raise ValueError("vector length must match the number of matrix columns")
    return tuple(
        sum(float(value) * float(vector[index]) for index, value in enumerate(row))
        for row in my_mx
    )


matrix = ((1, 2, 3), (4, 5, 6))
vector = (7, 8, 9)
print(mx_vec_mult(matrix, vector))

try:
    print(mx_vec_mult(((1, 2), (3, 4)), (5,)))
except ValueError as error:
    print("维度错误：", error)

try:
    print(mx_vec_mult(((1, "x"), (3, 4)), (5, 6)))
except TypeError as error:
    print("类型错误：", error)

try:
    print(mx_vec_mult(((1, 2), (3, 4)), [5, 6]))
except TypeError as error:
    print("类型错误：", error)
```

## 向量练习结果

- `vec_scalar_mult` 将标量分别乘到元组中的每个数值上，并返回同长度元组。
- `vec_inner_product` 将两个等长向量对应元素相乘后求和；输入类型错误时引发 `TypeError`，长度不同时引发 `ValueError`。
- `mx_vec_mult` 对矩阵的每一行与向量计算内积，返回由各行结果组成的元组；向量长度必须等于矩阵列数。

# 四、向量

1. 定义一个名为 `Vector` 的类。每个向量都有一个维度（非负整数）和一个元素列表或元组。初始化函数的第一个参数是维度，第二个参数是由整数或浮点数组成的列表或元组，表示向量的元素。如果只提供维度而不提供元素，请创建对应维度的零向量。若维度类型不正确或为负数，应抛出合适的异常；若维度与元素个数不一致，也应抛出异常。
2. 维度和元素应是类属性还是实例属性？说明选择的理由。
3. 实现向量对象的比较运算符，包括相等、小于、小于或等于、大于和大于或等于。两个 `Vector` 对象的维度相同，则比较坐标；否则按 Python 元组的字典序比较，即先比较第一个坐标，再比较第二个坐标，依此类推。例如，二维向量 `(2, 4)` 小于 `(2, 5)`。比较维度不同的两个向量时，应抛出异常。
4. 实现 `Vector.dot` 方法，接收另一个 `Vector` 对象，返回两个向量的内积。若参数类型不正确或维度不一致，应抛出合适的异常。
5. 实现 `Vector` 的乘法运算符。标量左乘或右乘（如 `2 * v` 和 `v * 2`）应返回一个各元素都乘以该标量的新 `Vector` 对象。当两个 `Vector` 对象相乘（`v * w`）时，对应位置的元素逐一相乘，结果仍是一个新 `Vector` 对象；例如，`Vector(2, (2, 3)) * Vector(2, (4, 5))` 的元素为 `(8, 15)`。若两个向量的维度不一致，应抛出合适的异常。

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if type(dimension) is not int:
            raise TypeError("dimension must be an integer")
        if dimension < 0:
            raise ValueError("dimension cannot be negative")

        if elements is None:
            elements = (0,) * dimension
        elif not isinstance(elements, (list, tuple)):
            raise TypeError("elements must be a list or tuple")

        if any(type(value) not in (int, float) for value in elements):
            raise TypeError("vector elements must be integers or floats")
        if len(elements) != dimension:
            raise ValueError("dimension must match the number of elements")

        self.dimension = dimension
        self.elements = tuple(elements)

    def _check_same_dimension(self, other):
        if not isinstance(other, Vector):
            raise TypeError("operand must be a Vector")
        if self.dimension != other.dimension:
            raise ValueError("vectors must have the same dimension")

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_same_dimension(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_same_dimension(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_same_dimension(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_same_dimension(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_same_dimension(other)
        return self.elements >= other.elements

    def dot(self, other):
        self._check_same_dimension(other)
        return sum(left * right for left, right in zip(self.elements, other.elements))

    def __mul__(self, other):
        if isinstance(other, Vector):
            self._check_same_dimension(other)
            return Vector(
                self.dimension,
                tuple(left * right for left, right in zip(self.elements, other.elements)),
            )
        if type(other) in (int, float):
            return Vector(
                self.dimension,
                tuple(value * other for value in self.elements),
            )
        return NotImplemented

    def __rmul__(self, other):
        return self.__mul__(other)


# 基本用法与关键运算
v = Vector(2, [2, 3])
w = Vector(2, (4, 5))
zero_vector = Vector(3)

print("v =", v)
print("默认零向量 =", zero_vector)
print("v == Vector(2, (2, 3)):", v == Vector(2, (2, 3)))
print("Vector(2, (2, 4)) < Vector(2, (2, 5)):", Vector(2, (2, 4)) < Vector(2, (2, 5)))
print("v.dot(w) =", v.dot(w))
print("2 * v =", 2 * v)
print("v * 2 =", v * 2)
print("v * w =", v * w)

assert v == Vector(2, (2, 3))
assert Vector(2, (2, 4)) < Vector(2, (2, 5))
assert v.dot(w) == 23
assert 2 * v == Vector(2, (4, 6))
assert v * 2 == Vector(2, (4, 6))
assert v * w == Vector(2, (8, 15))
assert zero_vector == Vector(3, (0, 0, 0))


def check_raises(exception_type, operation):
    try:
        operation()
    except exception_type:
        return
    raise AssertionError(f"expected {exception_type.__name__}")


check_raises(TypeError, lambda: Vector(2.0, (1, 2)))
check_raises(ValueError, lambda: Vector(-1))
check_raises(ValueError, lambda: Vector(2, (1,)))
check_raises(TypeError, lambda: Vector(2, (1, "2")))
check_raises(ValueError, lambda: Vector(1, (1,)) == Vector(2, (1, 0)))
check_raises(ValueError, lambda: Vector(1, (1,)).dot(Vector(2, (1, 0))))
check_raises(TypeError, lambda: Vector(1, (1,)).dot((1,)))
check_raises(ValueError, lambda: Vector(1, (1,)) * Vector(2, (1, 0)))
print("异常检查全部通过")
```

## 实现说明与结果

- `dimension` 和 `elements` 都是实例属性：每个向量可以有不同的维度和坐标，因此它们描述的是具体对象，而不是所有 `Vector` 对象共享的类属性。坐标在初始化时统一保存为元组，避免外部列表修改后悄悄改变向量内容。
- 只提供维度时，创建该维度的零向量，例如 `Vector(3)` 等价于 `Vector(3, (0, 0, 0))`。维度必须是非负整数；元素必须来自列表或元组、数量必须与维度一致，且每个元素必须是整数或浮点数。
- 比较运算先检查两个对象都是同维度的 `Vector`，再直接比较坐标元组。元组比较遵循字典序，所以 `(2, 4) < (2, 5)`；不同维度无法按本题规则比较，会引发 `ValueError`。
- `dot` 计算对应坐标乘积之和。例如 `(2, 3) · (4, 5) = 2×4 + 3×5 = 23`。`*` 支持标量左乘和右乘，也支持两个同维向量逐坐标相乘；因此 `(2, 3) * (4, 5) = (8, 15)`。这些运算都返回新的向量对象（内积除外，它返回数值）。
- 参数或元素类型错误时引发 `TypeError`；维度为负、元素数量不符或两个向量维度不一致时引发 `ValueError`。代码单元格末尾的断言覆盖了示例结果和这些错误情况。
