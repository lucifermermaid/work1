# Python 正则表达式练习

本部分主要练习正则表达式的完整匹配、分组、查找替换和边界约束。为便于直接运行，每个题目都封装为独立函数，并在下方给出典型输入及断言测试。

## 涵盖内容

- 字符串是否由字母和数字组成
- `a` 后接零个或多个、一个或多个 `b`
- 字符串中第一个 `a` 到最后一个 `b`
- IP 地址前导零和空格下划线互换
- 日期格式转换与特定单词查找
- 去除数字字母以外的字符和括号内容

```python
import re

def is_alphanumeric(text):
    return isinstance(text, str) and re.fullmatch(r"[A-Za-z0-9]+", text) is not None

def match_a_followed_by_zero_or_more_b(text):
    return isinstance(text, str) and re.fullmatch(r"ab*", text) is not None

def match_a_followed_by_one_or_more_b(text):
    return isinstance(text, str) and re.fullmatch(r"ab+", text) is not None

def match_a_to_b(text):
    return isinstance(text, str) and re.fullmatch(r"a[\s\S]*b", text) is not None

def remove_ip_leading_zero(ip_address):
    if not isinstance(ip_address, str):
        raise TypeError("ip_address 必须是字符串")
    return ".".join(str(int(segment)) if segment.isdigit() else segment for segment in ip_address.split("."))

def swap_spaces_and_underscores(text):
    if not isinstance(text, str):
        raise TypeError("text 必须是字符串")
    return re.sub(r"[ _]", lambda match: "_" if match.group(0) == " " else " ", text)

def convert_date_format(text):
    if not isinstance(text, str):
        raise TypeError("text 必须是字符串")
    match = re.fullmatch(r"(\d{4})-(\d{2})-(\d{2})", text)
    if match is None:
        return text
    year, month, day = match.groups()
    return f"{day}-{month}-{year}"

def find_words_starting_with_a_or_e(text):
    if not isinstance(text, str):
        raise TypeError("text 必须是字符串")
    return re.findall(r"\b[aeAE][A-Za-z]*\b", text)

def abbreviate_road(text):
    if not isinstance(text, str):
        raise TypeError("text 必须是字符串")
    return re.sub(r"\bRoad\b", "Rd", text)

def find_words_at_least_four_characters(text):
    if not isinstance(text, str):
        raise TypeError("text 必须是字符串")
    return re.findall(r"\b[A-Za-z0-9_]{4,}\b", text)

def remove_non_alphanumeric(text):
    if not isinstance(text, str):
        raise TypeError("text 必须是字符串")
    return re.sub(r"[^A-Za-z0-9]", "", text)

def remove_parentheses_and_content(text):
    if not isinstance(text, str):
        raise TypeError("text 必须是字符串")
    return re.sub(r"\s*\([^()]*\)", "", text).strip()
```

```python
print("第 1 题 Abc123:", is_alphanumeric("Abc123"))
print("第 1 题 Abc-123:", is_alphanumeric("Abc-123"))
print("第 1 题 空字符串:", is_alphanumeric(""))

print("第 2 题 a:", match_a_followed_by_zero_or_more_b("a"))
print("第 2 题 abbb:", match_a_followed_by_zero_or_more_b("abbb"))
print("第 2 题 ba:", match_a_followed_by_zero_or_more_b("ba"))

print("第 3 题 ab:", match_a_followed_by_one_or_more_b("ab"))
print("第 3 题 abbb:", match_a_followed_by_one_or_more_b("abbb"))
print("第 3 题 a:", match_a_followed_by_one_or_more_b("a"))

print("第 4 题 ab:", match_a_to_b("ab"))
print("第 4 题 a-middle-b:", match_a_to_b("a-middle-b"))
print("第 4 题 a-middle-c:", match_a_to_b("a-middle-c"))

print("第 5 题 192.168.001.010:", remove_ip_leading_zero("192.168.001.010"))
print("第 5 题 000.000.000.000:", remove_ip_leading_zero("000.000.000.000"))

print("第 6 题:", swap_spaces_and_underscores("Hello world_test"))
print("第 7 题:", convert_date_format("2026-11-24"))
print("第 8 题:", find_words_starting_with_a_or_e("Apple and Elephant eat easy"))
print("第 9 题:", abbreviate_road("100 Main Road"))
print("第 10 题:", find_words_at_least_four_characters("This is four"))
print("第 11 题:", remove_non_alphanumeric("Hello, W3$! world"))

samples = ["example (.com)", "w3resource", "github (.com)", "stackoverflow (.com)"]
result = [remove_parentheses_and_content(item) for item in samples]
print("第 12 题:")
for item in result:
    print(item)

assert is_alphanumeric("Abc123")
assert not is_alphanumeric("Abc-123")
assert match_a_followed_by_zero_or_more_b("abbb")
assert not match_a_followed_by_one_or_more_b("a")
assert match_a_to_b("a-middle-b")
assert remove_ip_leading_zero("192.168.001.010") == "192.168.1.10"
assert swap_spaces_and_underscores("Hello world_test") == "Hello_world test"
assert convert_date_format("2026-11-24") == "24-11-2026"
assert find_words_starting_with_a_or_e("Apple and Elephant eat easy") == ["Apple", "and", "Elephant", "eat", "easy"]
assert abbreviate_road("100 Main Road") == "100 Main Rd"
assert find_words_at_least_four_characters("This is four") == ["This", "four"]
assert remove_non_alphanumeric("Hello, W3$! world") == "HelloW3world"
assert result == ["example", "w3resource", "github", "stackoverflow"]
print("正则表达式测试全部通过")
```

# 从网页检索气候数据

使用 `requests` 获取网页，使用 `BeautifulSoup` 和 `html.parser` 定位气候数据表格。

## 实现思路

1. **获取网页**：发送 HTTP 请求，并设置请求超时和浏览器请求头。
2. **定位表格**：检查表格标题、表头或附近标题中是否包含 `Climate data for`。
3. **观察表格结构**：统计直接标签子节点的数量和标签名。
4. **提取行名**：优先读取带 `scope="row"` 的表头，其余情况取该行的第一个表头。
5. **过滤无关内容**：跳过空行、月份标题、数据来源及 `References` 等非气候指标。

> 网络不可用、页面结构变化或访问受限时，相关函数可能抛出 `requests.RequestException`；下方代码会捕获并显示错误。Notebook 末尾还提供了不依赖网络的结构测试。

## 测试城市

- Los Angeles
- Hyderabad
- Boston

```python
import re
from datetime import date

import requests
from bs4 import BeautifulSoup, Tag


def normalize_table_text(value):
    text = value.get_text(" ", strip=True)
    return re.sub(r"\s+", " ", text).strip()


def observe_table_structure(table):
    if not isinstance(table, Tag):
        return {"direct_tag_count": 0, "direct_tag_names": []}

    direct_tag_children = [
        child for child in table.children
        if isinstance(child, Tag)
    ]
    return {
        "direct_tag_count": len(direct_tag_children),
        "direct_tag_names": [child.name for child in direct_tag_children],
    }


def is_climate_table(table):
    caption = table.find("caption")
    if caption is not None and "climate data for" in normalize_table_text(caption).lower():
        return True

    for heading in table.find_all(["th", "td"]):
        if re.search(r"climate data for", normalize_table_text(heading), re.I):
            return True

    for heading in table.find_all_previous(["caption", "h1", "h2", "h3", "h4"]):
        if re.search(r"climate data for", normalize_table_text(heading), re.I):
            return True

    return False


def retrieve_climate_table(s):
    if not isinstance(s, str):
        raise TypeError("s 必须是 URL 字符串")

    response = requests.get(
        s,
        timeout=20,
        headers={"User-Agent": "Mozilla/5.0"}
    )
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")
    for table in soup.find_all("table"):
        if is_climate_table(table):
            return table

    return None


def list_climate_table_row_names(s):
    table = retrieve_climate_table(s)
    if table is None:
        return None

    row_names = []
    for tr in table.find_all("tr"):
        row_header = tr.find("th", attrs={"scope": "row"})
        if row_header is None:
            row_header = tr.find("th")

        if row_header is None:
            continue

        name = normalize_table_text(row_header)
        lower_name = name.lower()

        if not name:
            continue
        if "climate data for" in lower_name:
            continue
        if lower_name in {"month", "year", "metric", "source", "data source", "references"}:
            continue
        if re.search(r"^(data source|source|references)\b", lower_name):
            continue
        if tr.find("td") is None:
            continue
        if name not in row_names:
            row_names.append(name)

    return row_names
```

```python
los_angeles_url = "https://en.wikipedia.org/wiki/Los_Angeles"
hyderabad_url = "https://en.wikipedia.org/wiki/Hyderabad"
boston_url = "https://en.wikipedia.org/wiki/Boston"
access_date = date.today().isoformat()

print("Los Angeles URL:", los_angeles_url)
print("Hyderabad URL:", hyderabad_url)
print("Boston URL:", boston_url)
print("访问日期:", access_date)

for url in (los_angeles_url, hyderabad_url, boston_url):
    try:
        table = retrieve_climate_table(url)
        print(url, "气候表找到:", table is not None)
        if table is not None:
            print("行名:", list_climate_table_row_names(url))
            structure = observe_table_structure(table)
            print("直接子标签数量:", structure["direct_tag_count"])
            print("直接子标签名:", structure["direct_tag_names"])
    except requests.RequestException as error:
        print(url, "网络或 HTTP 错误:", error)
```

```python
# 不依赖网络的结构测试
test_html = """
<html>
  <body>
    <h2>Climate data for Example City</h2>
    <table>
      <caption>Climate data for Example City</caption>
      <thead><tr><th scope="col">Month</th><th scope="col">Value</th></tr></thead>
      <tbody>
        <tr><th scope="row">Average high °C</th><td>20</td></tr>
        <tr><th scope="row">Average low °C</th><td>10</td></tr>
        <tr><th scope="row">Average precipitation mm</th><td>50</td></tr>
        <tr><th scope="row">Source</th><td>Example</td></tr>
      </tbody>
    </table>
  </body>
</html>
"""
test_soup = BeautifulSoup(test_html, "html.parser")
test_table = test_soup.find("table")
assert is_climate_table(test_table)
print("气候表结构测试:", is_climate_table(test_table))
structure = observe_table_structure(test_table)
print("直接子标签数量:", structure["direct_tag_count"])
print("直接子标签名:", structure["direct_tag_names"])

synthetic_names = []
for tr in test_table.find_all("tr"):
    header = tr.find("th", attrs={"scope": "row"})
    if header is not None and tr.find("td") is not None:
        name = normalize_table_text(header)
        if name.lower() not in {"month", "source", "data source", "references"}:
            synthetic_names.append(name)

assert synthetic_names == ["Average high °C", "Average low °C", "Average precipitation mm"]
print("行名结构测试:", synthetic_names)
```
