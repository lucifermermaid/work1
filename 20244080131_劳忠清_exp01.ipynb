### 3. 类型转换与异常

`int("123")` 将数字文本转为整数，`float("3.14")` 转为浮点数，`int(3.9)` 会截去小数部分而得到 `3`（不是四舍五入）。`int("goat")` 失败是因为该文本不是合法的整数表示，因此捕获并显示 `ValueError`。

```python
converted_integer = int("123")
converted_float = float("3.14")
truncated_integer = int(3.9)

print(f'int("123") = {converted_integer}，类型：{type(converted_integer).__name__}')
print(f'float("3.14") = {converted_float}，类型：{type(converted_float).__name__}')
print(f"int(3.9) = {truncated_integer}，小数部分被截去")

try:
    int("goat")
except ValueError as error:
    print(f'int("goat") 引发 {type(error).__name__}：文本不是合法的整数表示。')
```

### 5. 字符串拼接、重复与不可变性

`+` 用于拼接字符串，`*` 用于重复字符串。Python 字符串不可变，不能直接给某个字符赋新值；可通过切片和拼接构造新字符串。

```python
word = "Python"
print(f'拼接：{word + " Data"}')
print(f"重复：{word * 3}")

try:
    word[0] = "J"
except TypeError as error:
    print(f"直接修改单个字符失败，得到 {type(error).__name__}：字符串不可变。")

new_word = "J" + word[1:]
print(f"用切片和拼接构造的新字符串：{new_word}")
print(f"原字符串仍为：{word}")
```

## 实验2：函数、条件、循环与递归



### 1. 成绩等级函数

按分数区间返回等级；先检查范围，低于 0 分或高于 100 分时抛出 `ValueError`。

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("score 必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for score in [100, 90, 89, 80, 79, 60, 59, 0]:
    print(f"{score} 分：{score_level(score)}")

for invalid_score in [-1, 101]:
    try:
        score_level(invalid_score)
    except ValueError as error:
        print(f"{invalid_score} 分：捕获 {type(error).__name__}（{error}）")
```

### 2. 使用 for 和 while 求和

分别使用两种循环累加 1 到 100，并比较结果是否相同。

```python
for_sum = 0
for number in range(1, 101):
    for_sum += number

while_sum = 0
number = 1
while number <= 100:
    while_sum += number
    number += 1

print(f"for 循环计算的 1 到 100 的和：{for_sum}")
print(f"while 循环计算的 1 到 100 的和：{while_sum}")
print(f"两种结果相同：{for_sum == while_sum}")
```

### 3. 递归版与迭代版阶乘

两种函数都要求输入为非负整数；递归版以 `0! = 1` 为基准，迭代版用循环累乘。比较 `n = 0、1、5、10` 的结果。

```python
def validate_factorial_input(n):
    if isinstance(n, bool) or not isinstance(n, int):
        raise TypeError("n 必须是整数")
    if n < 0:
        raise ValueError("n 必须是非负整数")


def factorial_recursive(n):
    validate_factorial_input(n)
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    validate_factorial_input(n)
    result = 1
    for factor in range(2, n + 1):
        result *= factor
    return result


for n in [0, 1, 5, 10]:
    recursive_result = factorial_recursive(n)
    iterative_result = factorial_iterative(n)
    print(
        f"n={n}: 递归={recursive_result}，迭代={iterative_result}，"
        f"结果相同={recursive_result == iterative_result}"
    )
```

### 4. 列表统计 summary(values)

函数接收非空数值列表，返回 `(最小值, 最大值, 平均值)`；调用时用三个变量进行元组解包。

```python
def summary(values):
    if not values:
        raise ValueError("values 不能为空")
    if any(isinstance(value, bool) or not isinstance(value, (int, float)) for value in values):
        raise TypeError("values 必须只包含数值")
    return min(values), max(values), sum(values) / len(values)


sample_values = [3, 7, 2, 10, 8]
minimum, maximum, average = summary(sample_values)
print(f"数据：{sample_values}")
print(f"最小值={minimum}，最大值={maximum}，平均值={average}")
```

### 5. 可变参数求和 sum_all(*args)

`*args` 将任意数量的位置参数收集成元组；用 `sum()` 求和，无参数时结果为 `0`。

```python
def sum_all(*args):
    return sum(args)


print(f"sum_all() = {sum_all()}")
print(f"sum_all(5) = {sum_all(5)}")
print(f"sum_all(1, 2, 3, 4) = {sum_all(1, 2, 3, 4)}")
```

## 实验3：列表、字典、元组与基础统计



```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

score_by_name = dict(zip(names, scores))
print("姓名与成绩（items）：")
for name, score in score_by_name.items():
    print(f"{name}: {score}")

print("\n序号、姓名与成绩（enumerate 从 1 开始）：")
for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"{index}. {name}: {score}")

passed_names = [name for name, score in zip(names, scores) if score >= 60]
squared_scores = [score ** 2 for score in scores]
print(f"及格学生：{passed_names}")
print(f"成绩平方：{squared_scores}")

student_count = len(scores)
average_score = sum(scores) / student_count
highest_score = max(scores)
lowest_score = min(scores)
pass_rate = sum(score >= 60 for score in scores) / student_count
print(f"人数：{student_count}")
print(f"平均分：{average_score:.2f}")
print(f"最高分：{highest_score}")
print(f"最低分：{lowest_score}")
print(f"及格率：{pass_rate:.2%}")

original_demo = scores.copy()
alias = original_demo
copied = original_demo.copy()
alias[0] = 0
print(f"修改 alias 后，原列表：{original_demo}")
print(f"修改 alias 后，副本：{copied}")

student_record = ("Alice", 85)
print(f"学生元组：{student_record}，长度：{len(student_record)}")
try:
    student_record[1] = 90
except TypeError as error:
    print(f"元组元素不能重新赋值：{type(error).__name__}")

unequal_zip = list(zip(["A", "B", "C"], [1, 2]))
print(f"长度不同的序列 zip 结果：{unequal_zip}")
```

## 实验5：类、继承与对象复制

从实验 4 的有效记录创建学生对象，再演示继承和浅/深复制对嵌套列表的影响。

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (
            f"GraduateStudent(name={self.name}, score={self.score}, "
            f"research_topic={self.research_topic})"
        )


student_objects = [
    Student(record["name"], record["score"])
    for record in valid_records
]
for student in student_objects:
    print(f"{student}，及格：{student.is_passed()}")

graduate = GraduateStudent("Iris", 88, "Data Analysis")
print(f"继承类对象：{graduate}，及格：{graduate.is_passed()}")
```
